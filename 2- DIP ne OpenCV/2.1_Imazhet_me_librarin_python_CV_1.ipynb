### Imazhe në Shkallë Gri

Imazhet gri (grayscale) kanë vlera pikseli që përfaqësojnë sasinë e dritës ose intensitetin. Hijet më të lehta të gri kanë intensitet më të lartë, ndërsa hijet më të errëta kanë intensitet më të ulët. E bardha ka intensitetin më të lartë dhe e zeza më të ulët. Ne mund ta konvertojmë një imazh në Gri (Gray Scale) duke përdorur një kod për konvertimin e ngjyrave dhe funksionin `cvtColor`.

Kodi për të konvertuar nga RGB në gri është **cv2.COLOR_BGR2GRAY**. Ne e aplikojmë funksionin cvtColor në këtë mënyrë:

````python
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
````

Matrica e imazhit ka vetëm dy dimensione, domethënë ka vetëm një kanal ngjyre.

````python
image_gray.shape
````

Mund të vizualizojmë imazhin duke përdorur `imshow`, por duhet të specifikojmë që harta e ngjyrave të jetë gri:

````python
plt.figure(figsize=(10, 10))
plt.imshow(image_gray, cmap='gray')
plt.show()
````

Mund ta ruajmë imazhin si një imazh në shkallë gri, ta ruajmë gjithashtu si një `jpg`, në direktorinë tonë të punës.

````python
cv2.imwrite('lena_gray_cv.jpg', image_gray)
````

Mund të ngarkoni gjithashtu një imazh në shkallë gri, duhet të vendosim parametrin <code>flag</code> në kodin për konvertimin në ngjyrë gri: <code>cv2.COLOR_BGR2GRAY</code>:

````python
im_gray = cv2.imread('barbara.png', cv2.IMREAD_GRAYSCALE)
````

Ne mund te plotojme imazhin:

````python
plt.figure(figsize=(10,10))
plt.imshow(im_gray,cmap='gray')
plt.show()
````

### Kanalet e Ngjyrave

Ne mund të punojmë gjithashtu me kanale të ndryshme ngjyrash. Merrni parasysh imazhin e mëposhtëm:

````python
baboon=cv2.imread('baboon.png')
plt.figure(figsize=(10,10))
plt.imshow(cv2.cvtColor(baboon, cv2.COLOR_BGR2RGB))
plt.show()
````

Ne mund të marrim ngjyrat e ndryshme RGB dhe t'i caktojmë ato në variablat <code>blue</code>, <code>green</code>, dhe <code>red</code>, në formatin (B, G, R).

````python
blue, green, red = baboon[:, :, 0], baboon[:, :, 1], baboon[:, :, 2]
````

Ne mund të bashkojmë çdo kanal imazhi duke përdorur funksionin <code>vconcat</code>.

````python
im_bgr = cv2.vconcat([blue, green, red])
````

Kur plotojmë imazhin me ngjyrë pranë kanalit të kuq në shkallë gri, shohim se zonat me ngjyrë të kuqe kanë vlera më të larta intensiteti.

````python
plt.figure(figsize=(10,10))
plt.subplot(121)
plt.imshow(cv2.cvtColor(baboon, cv2.COLOR_BGR2RGB))
plt.title("RGB image")
plt.subplot(122)
plt.imshow(im_bgr,cmap='gray')
plt.title("Different color channels  blue (top), green (middle), red (bottom)  ")
plt.show()
````

Nëse dëshironi të rishpërndani një array në një variabël tjetër, duhet të përdorni metodën `copy` (ne do ta trajtojmë këtë në seksionin e ardhshëm).

````python
A = new_image.copy()
plt.imshow(A)
plt.show()
````

Nëse nuk aplikohet metoda `copy()`, variabla do të tregojë në të njëjtën vendndodhje në memorje. Konsideroni variablën `B` më poshtë, nëse vendosim të gjithë vlerat e array-t `A` në zero, pasi `A` dhe `B` tregojnë në të njëjtin objekt në memorje, `B` do të ketë gjithashtu elementët zero:

````python
B = A
A[:,:,:] = 0
plt.imshow(B)
plt.show()
````
