## Decompozimi i Vlerës Singullare (SVD) për Imazhet

Ne mund të aplikojmë algoritme të dizajnuara për matrica. Mund të përdorim Decompozimin e Vlerës Singullare (SVD), duke e dekompozuar matricën e imazhit në një produkt të tre matricave.



Ky kod dekompozon një imazh bardh e zi (**grayscale**) në tre matrica:

- **$U$** → Matricë ortogonale që përmban "bazat e kolonave" të matricës origjinale  
- **$s$** → Vektori që përmban **vlerat singullare**, që janë si "peshat" kryesore të matricës  
- **$V$** → Matricë ortogonale që përmban "bazat e rreshtave" të matricës origjinale  

#### Shpjegimi
SVD dekompozon një matricë $A$ (në këtë rast, matricën e imazhit bardh e zi) në formën:

$$
A = U \cdot S \cdot V^T
$$

Ku:  
- **$U$ dhe $V$ janë matrica ortogonale**  
- **$S$ është matricë diagonale** që përmban **vlerat singullare** të matricës  

Nëse **`im_gray`** është një matricë me përmasa $m \times n$, atëherë:

- **$U$** do të jetë $m \times m$  
- **$s$** do të jetë një **vektor me vlerat singullare** (jo një matricë)  
- **$V$** do të jetë $n \times n$  


```python
U, s, V = np.linalg.svd(im_gray , full_matrices=True)
```

```python
plt.imshow(U,cmap='gray')
plt.show()
print(s.shape, V.shape)
```

Ne shohim që <code>s</code> nuk është drejtkëndëshe:

```python
s.shape
```

Ne mund ta konvertojmë <code>s</code> në një matricë diagonale <code>S</code>:

```python
S = np.zeros((im_gray.shape[0], im_gray.shape[1]))
S[:im_gray.shape[0], :im_gray.shape[0]] = np.diag(s)
```

```python
print(np.max(S))
```

```python
plt.imshow(S,cmap='gray')
plt.show()
```

Ne mund të shfaqim matricat U dhe V:

```python
plot_image(U, V, title_1="Matrix U", title_2="Matrix V")
```

Ne shohim që shumica e elementëve në S janë zero:

```python
S
```

```python
plt.imshow(S, cmap='gray')
plt.show()
```

Ne mund të gjejmë produktin e matricave të të gjitha matricave. Së pari, mund të kryejmë shumëzimin e matricave mes S dhe U dhe ta caktojmë në `B` dhe të shfaqim rezultatet:

```python
B = S.dot(V)
plt.imshow(B,cmap='gray')
plt.show()
```

Ne mund të gjejmë produktin e matricave të `U`, `S`, dhe `B`. Ne shohim që kjo është e gjithë pamja e imazhit:

```python
A = U.dot(B)
```

```python
plt.imshow(A,cmap='gray')
plt.show()
```

Doli që shumë elementë janë të panevojshëm. Ne mund të eleminojmë disa rreshta dhe kolona nga S dhe V dhe të afrojmë imazhin duke gjetur produktin:

```python
for n_component in [1,10,100,200, 500]:
    S_new = S[:, :n_component]
    V_new = V[:n_component, :]
    A = U.dot(S_new.dot(V_new))
    plt.imshow(A,cmap='gray')
    plt.title("Number of Components:"+str(n_component))
    plt.show()
```

Ne shohim që na duhen vetëm 100 deri në 200 Komponentë për të përfaqësuar imazhin.
