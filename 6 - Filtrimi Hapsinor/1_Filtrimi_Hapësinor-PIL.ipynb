## Filtrimi Linear


Filtrimi përfshin përmirësimin e një imazhi, për shembull, heqjen e zhurmës nga një imazh.

Zhurma mund të shkaktohet nga një kamerë e dobët ose nga një kompresim i keq i imazhit.
Të njëjtët faktorë që shkaktojnë zhurmën mund të çojnë edhe në imazhe të turbullta. Ne mund të aplikojmë filtra për t’i mprehur këto imazhe.

Konvolucioni është një mënyrë standarde për të filtruar një imazh. Filtri quhet bërthamë (kernel) dhe bërthama të ndryshme kryejnë detyra të ndryshme. Për më tepër, konvolucioni përdoret në shumë nga algoritmet më të avancuara të inteligjencës artificiale.

Thjesht marrim prodhimin skalar midis bërthamës dhe një pjese të imazhit me të njëjtën madhësi. Më pas, zhvendosim bërthamën dhe e përsërisim procesin.

Zgjedhim imazhin e meposhtem per te kryer ptoceset e permendura me heret:

**Procesi i aplikimit (konvolucion):**

Kernel-i vendoset mbi një zonë 3x3 të imazhit.

Bëhet shumëzim element më element (pikselët e imazhit × vlerat e kernel-it).

Mblidhen të gjitha vlerat e shumëzuara.

Vlera përfundimtare vendoset në pikselin qendror të asaj zone.

Kernel-i lëviz një hap përpara dhe e përsërit procesin.

```python
# ngarkojme imazhin nga pathi perkates
image = Image.open("lenna.png")
# lexojme imazhin dhe e shfaqim ate
plt.figure(figsize=(5,5))
plt.imshow(image)
plt.show()
```

Imazhet me të cilat po punojmë janë të përbëra nga vlera RGB, të cilat janë vlera nga 0 deri në 255. Nese kemi zhurma ne imazh, zero do të thotë zhurmë e bardhë(grainy -, efect).

```python
# Marrim numrin e rreshtave dhe kolonave në imazh
rows, cols = image.size
#* Krijojm vlera duke përdorur një shpërndarje normale me mesataren 0 dhe devijimin standard 15,
#* vlerat konvertohen në unit8, që do të thotë se vlerat janë midis 0 dhe 255
noise = np.random.normal(0, 15, (rows, cols, 3)).astype(np.uint8)
# Shtojm zhurmën në imazh
noisy_image = image + noise
# Krijon një imazh PIL nga një array
noisy_image = Image.fromarray(noisy_image)
# Shfaqim imazhin origjinal dhe imazhin me zhurmë duke përdorur funksionin e definuar më sipër
plot_image(image, noisy_image, title_1="Imazhi Origjinali", title_2="Imazhi me Zhurmë")
```

Kur shtojmë zhurmë në një imazh, ndonjëherë vlera mund të jetë më e madhe se 255, në këtë rast, 256 hiqet nga vlera për ta kthyer numrin dhe për ta mbajtur atë mes 0 dhe 255.

Për shembull, marrim një imazh me një vlerë RGB prej 137 dhe shtojmë zhurmë me një vlerë RGB prej 215 për të marrë një vlerë RGB prej 352. Ne pastaj heqim 256, numrin total të mundshëm të vlerave mes 0 dhe 255, për të marrë një numër mes 0 dhe 255.

### Filtrimi i Zhurmës


Për të krijuar kernela(berthama/filtra) të personalizuara dhe për të përdorur filtrat e parapërcaktuar, duhet të importojmë bibliotekën vijuese:


```python
from PIL import ImageFilter
```

Filtrat e zbutjes mesatarizojnë pikselët brenda një rajoni, dhe ndonjëherë quhen filtra të kalimit të ulët. Për filtrimin me mesatare, kernel-i thjesht bën mesataren e pikselëve në një rajon/lagje.


```python
#Krijojm një kernel që është një matric 5x5 ku çdo vlerë është 1/36
kernel = np.ones((5, 5)) / 36
print(kernel.shape)
print(kernel)
#Krijojm një filter të imazhit duke siguruar madhësinë e kernel-it
kernel_filter = ImageFilter.Kernel((5, 5), kernel.flatten())
```

Funksioni <code>filter</code> kryen një konvolucion midis imazhit dhe kernel-it në çdo kanal ngjyre të pavarur.



```python
# Filtëron imazhet duke përdorur kernel-in.
image_filtered = noisy_image.filter(kernel_filter)
```

Ne mund të vizualizojmë imazhin para dhe pas filtrimit. Vërejmë se zhurma është reduktuar, por imazhi është bërë i paqartë(i turbullt).


```python
# Vizualizojm imazhin e filtruar dhe imazhin me zhurmë:
plot_image(image_filtered, noisy_image,title_1="Imazhi i Filtruar",title_2="Image me Zhurem")
```

#### Detyre
-> apliko histogram equalization ne cdo kanal te imazhit te filtruar te lenes edhe printoje imazhin e filtruar

Një kernel më i vogël e mban imazhin të mprehtë(te qarte), por filtron më pak zhurmë.
Kështu qe provojm një kernel 3x3. Mund të shihni që supet janë më të mprehta në këtë imazh, por zhurma e gjelbër është më e ndritshme sesa në imazhin e filtruar më lart.

```python
# Krijojm një filter(kernel) që është një matric 3x3 ku çdo vlerë është 1/36  
kernel = np.ones((3,3))/36  
# Krijojm një Kernel të ImageFilter duke ofruar madhësinë e kernel-it dhe vlerat e tij 
kernel_filter = ImageFilter.Kernel((3,3), kernel.flatten())  
# Filtrojm imazhet duke përdorur kernel-in  
image_filtered = noisy_image.filter(kernel_filter)  
# Përdorim funksionin e definuar më lart për të shfaqur imazhin e filtruar dhe imazhin me zhurmë  
plot_image(image_filtered, noisy_image,title_1="Imazhi i Filtruar",title_2="Imazhi me Zhurmë")
```

### Zbutja Gaussiane(Gaussian Blur)

Zbutja Gaussian është një teknikë filtrimi që përdor një kernel të shpërndarë nga një shpërndarje normale (shpërndarje Gaussian) për të zbutur imazhet. Ky filtër përdoret gjerësisht për të hequr zhurmën dhe për të krijuar një efekt të butë dhe të shkrirë. Zbatimi i zbutjes Gaussian ndihmon në zbutjen e skajeve dhe hollimin e detajeve të imazhit.

Në zbutjen Gaussian, vlerat e kernel-it varen nga shpërndarja normale, ku vlera më afër qendrës ka ndikim më të madh. Ky filtër është shumë i dobishëm për të pastruar imazhet nga zhurmat dhe për t'i përgatitur ato për analiza të mëtejshme, si detektimi i skajeve.

Për të kryer zbutjen Gaussian (Gaussian Blur), ne përdorim funksionin `filter` mbi një imazh duke përdorur filtrin e paracaktuar `ImageFilter.GaussianBlur`.

Parametrat

<p><code>radius</code>: rrezja e kernelit të blur-it, parazgjedhja automatike është 2</p>

```python
# Filtron imazhet duke përdorur GaussianBlur
image_filtered = noisy_image.filter(ImageFilter.GaussianBlur)
# Përdor funksionin për të shfaqur imazhin e filtruar dhe imazhin me zhurmë
plot_image(image_filtered , noisy_image, title_1="Imazhi i filtruar- GaussianBlur", title_2="Imazhi me Zhurmë")
```

```python
ii= np.array(image_filtered)
ii
```

Le të provojmë duke përdorur një kernel 4 me 4.

```python
# Filtron imazhet duke përdorur GaussianBlur në imazhin me zhurmë duke përdorur një kernel 4 me 4
image_filtered = noisy_image.filter(ImageFilter.GaussianBlur(4))
# Shfaq imazhin e filtruar dhe imazhin pa filtruar me zhurmë
plot_image(image_filtered , noisy_image,title_1="Imazhi i filtruar",title_2="Imazhi me zhurmë")
```

### Përforcimi i Imazhit (Image Sharpening)


Përforcimi(theksimi) i Imazhit përfshin zbutjen e imazhit dhe llogaritjen e derivateve. Ne mund ta realizojmë përforcimin e imazhit duke aplikuar filtrin e mëposhtëm.


```python
# Kernel i zakonshëm për përforcimin e imazhit
kernel = np.array([[-1,-1,-1], 
                   [-1, 9,-1],
                   [-1,-1,-1]])
kernel = ImageFilter.Kernel((3,3), kernel.flatten())
# Aplikon filtrin e përforcimit duke përdorur kernel-in në imazhin origjinal pa zhurmë
sharpened = image.filter(kernel)
# Ploton imazhin e përforcuar dhe imazhin origjinal pa zhurmë
plot_image(sharpened , image, title_1="Imazhi i përforcuar", title_2="Imazhi")

```

Ne mund të përforcojmë gjithashtu duke përdorur një filtrin e paracaktuar automatikishte ne kete funksion.


```python
# Përforcon imazhin duke përdorur filtër të paracaktuar nga PIL
sharpened = image.filter(ImageFilter.SHARPEN)
# Vizaton imazhin e përforcuar dhe imazhin origjinal pa zhurmë
plot_image(sharpened , image, title_1="Imazhi i përforcuar", title_2="Imazhi")
```

## Skajet/Kufijet


Kufijtë janë aty ku intensitetet e pikseleve ndryshojnë.

Gradienti i një funksioni tregon normën e ndryshimit; ne mund të aprogojmë/perafrojm gradintin e një imazhi të shkallës gri me konvolucion. Marrim parasysh imazhin e mëposhtëm:


```python
# Ngarkojm imazhin nga file perkates
img_gray = Image.open('barbara.png')
# shfaqim imazhin nga array i të dhënave, vëreni se është një matricë dy-dimensionale në vend që të ishte tre-dimensionale pasi nuk ka ngjyrë
plt.imshow(img_gray ,cmap='gray')
```

Përmirësojmë skajet që ato të kapen më mirë kur përdorim detektimin e skajeve.


```python
# Filtrojm imazhet duke përdorur filtrin EDGE_ENHANCE
img_gray = img_gray.filter(ImageFilter.EDGE_ENHANCE)
# Paraqesim imazhin e përmirësuar
plt.imshow(img_gray ,cmap='gray')
```

```python
# Filtrojm imazhet duke përdorur filtrin FIND_EDGES
img_gray = img_gray.filter(ImageFilter.FIND_EDGES)
# Paraqesim imazhin e filtruar
plt.figure(figsize=(10,10))
plt.imshow(img_gray ,cmap='gray')
```

## Filtrat Median

Filtrat median gjejnë medianën e të gjitha pikselëve nën zonën e kernelit(zona ku do vendoset) dhe elementi qendror zëvendësohet me këtë vlerë mediane.

Ne mund të aplikojmë filtrat median në imazhe të zakonshme, por le të shohim se si mund të përdorim një filtër median për të përmirësuar segmentimin.

Marrim si shembull imazhin e kameramanit: 


```python
# Ngarkojm imazhin e kameramanit
image = Image.open("cameraman.jpeg")
# Shfaqim imazhin
plt.figure(figsize=(10,10))

plt.imshow(image, cmap="gray")
```

Filtrimi me mediane thekson kufijtë ndërmjet objekteve dhe përmirëson ndarjen midis imazhit dhe sfondit. Kur përdoret për segmentim, ai ndihmon në largimin e zhurmës nga imazhi dhe bën që objektet, siç është figura e kameramanit, të duken më të ndara dhe më të qarta në sfondin e mbetur.

Kjo është veçanërisht e dobishme kur kemi zhurmë të vogël në imazhe, duke përmirësuar ndarjen e sfondit dhe objektit për një segmentim më të saktë.

```python
image = image.filter(ImageFilter.MedianFilter)
plt.figure(figsize=(10,10))

plt.imshow(image,cmap="gray")
```
